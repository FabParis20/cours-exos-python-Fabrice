### Q2: Generate Square Dictionary
Create a Python function that takes an integer ( n ) as input and generates a dictionary containing pairs ( (i, i^2) ) for all integers ( i ) from 1 to ( n ) (inclusive). The function should then return this dictionary.

Level: Beginner

````python
def generate_square_dict(nombre):
    square_dict = {}
    for chiffre in range(1, nombre + 1):
        carre = chiffre * chiffre
        square_dict[chiffre] = carre
    return square_dict
        
    
````

````python
print(generate_square_dict(4))
````

### Q3. Sequence to List and Tuple
Create a Python function that takes a sequence of comma-separated numbers as input and generates both a list and a tuple containing those numbers.
Level: Beginner

Input:

convert_input_to_list_and_tuple("3,6,5,3,2,8")
Expected Output:

(['3', '6', '5', '3', '2', '8'], ('3', '6', '5', '3', '2', '8'))

Hints:

- The input will be provided as a string.
- Use the split(",") method to convert the string into a list.
- The tuple() method can convert a list into a tuple.

````python
def cree_liste_tuple(sequence):
    # Sanity check
    if type(sequence) != str:
        print("La séquence doit être une suite de chiffres séparés par des virgules dans une chaine de caractères")
    else:
        liste = sequence.split(",")
        liste_propre = []
        # Sanity check bien un entier
        for car in liste:
            try:
                nb = int(car)
                liste_propre.append(nb)
            except:
                print("Les éléments de la séquences doivent être des entiers")
        return liste_propre, tuple(liste_propre)       
            
````

````python
print(cree_liste_tuple("1, 4, 5"))
````

````python
import math

def calcul(D):
    liste_result = []
    possibles_valeurs = D.split(",")
    for item in possibles_valeurs:
        convert = float(item)
        result = round(math.sqrt((2 * 50 * convert)/30))
        liste_result.append(result)
    return liste_result
        
````

````python
print(calcul("100, 150, 180"))
````
