# Grille de lecture — fonction train_model

## Niveaux d’imbrication
- **epoch loop** (`for epoch in range(num_epochs)`)
  - **phase loop** (`for phase in ['train', 'val']`)
    - **batch loop** (`for inputs, labels in dataloaders[phase]`)

## Variables clés
- `running_loss` : cumul de la perte sur un phase (train/val)
- `running_corrects` : cumul des prédictions correctes
- `best_acc` : meilleure accuracy validée rencontrée

## Entrées / sorties par niveau
- **Batch**  
  - Entrées : `inputs`, `labels`  
  - Sorties : `outputs`, `preds`, `loss`  

- **Phase (train/val)**  
  - Entrées : stats batch cumulées  
  - Sorties : `epoch_loss`, `epoch_acc`  

- **Epoch**  
  - Entrée : toutes les phases de l’époque  
  - Sorties : affichage des métriques + éventuelle mise à jour du `best_acc` et sauvegarde des poids

## Conditions importantes
- `if phase == 'train'`: active le mode entraînement, calcule les gradients, met à jour les poids  
- `if phase == 'val' and epoch_acc > best_acc`: sauvegarde le meilleur modèle

## Fin de fonction
- Affiche temps total et meilleure accuracy
- Recharge le meilleur modèle depuis `best_model_params_path`
